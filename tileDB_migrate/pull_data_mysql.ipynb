```python
import mysql.connector as connector
import numpy as np
import pandas as pd
import os

HEIGHT = 256
WIDTH = 384
```

```python
conn = connector.connect(
    user="root",
    password="",
    unix_socket="/tmp/mysql_database_dev.sock",
    database="heat_db",
)
```

```python
cur = conn.cursor(dictionary=True)
```

```python
def select_specific_pixel(x, y):
    """特定のピクセルの値を取得"""
    query = f"SELECT pixel FROM pix01 WHERE x = {x} AND y = {y}"
    cur.execute(query)
    result = cur.fetchall()
    return result
```

```python
specific_pixel = select_specific_pixel(23, 23)
print(f"特定ピクセルの値: {specific_pixel}")
```
